# Dataset exploration

This version reads the M5 `sales_train_evaluation.csv` file locally. It checks `data/raw/` first, then the project root, so it works with the CSV files currently copied into the repository root. Keep raw data files out of Git; this project ignores CSV files.

How many rows and columns are contained in the dataset?

The notebook finds the project data whether Jupyter starts in the project root or the `notebooks` folder. It prefers `data/raw/` when the expected file is there.

```python
from pathlib import Path
import pandas as pd

# Find the project root from the current working directory or its parents.
search_roots = (Path.cwd().resolve(), *Path.cwd().resolve().parents)
project_root = next(
    (root for root in search_roots if (root / "data" / "raw").is_dir()
     or (root / "sales_train_evaluation.csv").is_file()),
    None,
)
if project_root is None:
    raise FileNotFoundError("Could not locate the project data. Put the M5 CSV files in data/raw/ or the project root.")

raw_data_dir = project_root / "data" / "raw"
data_dir = raw_data_dir if (raw_data_dir / "sales_train_evaluation.csv").is_file() else project_root
sales_file = data_dir / "sales_train_evaluation.csv"
if not sales_file.is_file():
    raise FileNotFoundError(f"Expected {sales_file}. Add sales_train_evaluation.csv to data/raw/ or the project root.")

# Show which local M5 CSV files are present and where they were found.
csv_inventory = pd.DataFrame(
    [{"file": path.name, "size_mb": round(path.stat().st_size / 1_000_000, 2)}
     for path in sorted(data_dir.glob("*.csv"))]
)
print(f"Data folder: {data_dir}")
display(csv_inventory)

df = pd.read_csv(sales_file)
print(f"Loaded: {sales_file}")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

Is the dataset complete? i.e. no missing values?

```python
missing_values = df.isnull().sum()

print("Total missing values:", missing_values.sum())

print(df.isnull().values.any())
```

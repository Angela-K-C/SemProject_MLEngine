How many rows and columns are contained in the dataset?



```python
import pandas as pd

df = pd.read_csv("sales_train_evaluation.csv")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

What datatypes are contained in the dataset?

```python
print(df.dtypes.head(10))
```

Is the dataset complete? i.e. no missing values?

```python
missing_values = df.isnull().sum()

print("Total missing values:", missing_values.sum())

print(df.isnull().values.any())
```

Slice out the first 15 rows and last 20 rows and merge them into a new
dataframe titled “df_sample”. Display this new dataframe.

```python
first_15 = df.head(15)

last_20 = df.tail(20)

df_sample = pd.concat([first_15, last_20])

display(df_sample)
```
